## Explaining inference and training

ASTscribe analyzes the source strings below **without executing them**. PyTorch is not required.

```python
from astscribe import explain

inference = """import torch
model.eval()
with torch.no_grad():
    outputs = model(inputs)
"""
print(explain(inference, style="concise"))
```

```python
training = """model.train()
for batch in train_loader:
    optimizer.zero_grad()
    outputs = model(**batch)
    loss = outputs.loss
    loss.backward()
    optimizer.step()
"""
print(explain(training, style="concise"))
```

Inspect structured, evidence-backed inference flags instead of parsing the explanatory prose.

```python
from astscribe import analyze

result = analyze(inference)
print("Inference:", result.inference is not None)
print("No gradients:", result.inference.gradient_tracking_disabled if result.inference else False)
```
