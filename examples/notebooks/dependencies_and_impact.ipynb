## Notebook dependencies and impact

Analyze a realistic chain of cell definitions statically, then inspect the blast radius of a change.

```python
from astscribe import NotebookAnalyzer

notebook = NotebookAnalyzer.from_cells([
    "raw = 10",
    "features = raw * 2",
    "prediction = features + 1",
])
print(notebook.render_dependency_graph())
```

Which cells might need re-running when the input definition changes?

```python
print(notebook.render_impact(0))
```

```python
print(notebook.render_diagnostics())
```
